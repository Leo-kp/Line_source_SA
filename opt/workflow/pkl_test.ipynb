```python
import joblib
import numpy as np
import matplotlib.pyplot as plt
```

```python
pipeline = joblib.load("gp_pipeline_latest.pkl")

x_1 = np.linspace(10, 200, 200)
x_2 = np.linspace(10, 150, 200)
x_3 = np.linspace(0.1, 10, 200)
x_4 = np.linspace(20, 30, 200)
```

```python
pipeline
```

```python
X_grid = np.column_stack([x_1,x_2, x_3, x_4])
x_scaled = pipeline.name_steps["scaler"].transfomr(X_grid)
p_mean, p_std = pipeline.name_sted["gp"].predict(x_scaled, return_std = True)

```

```python
step_axis = np.arange(200)

plt.figure(figsize=(10,6))

plt.plot(step_axis, p_mean, "b--", lw=2, label= "GP Mean", alpha = 0.8)
plt.fill_between(
    step_axis, 
    p_mean - 1.96*p_std,
    p_mean + 1.96 * p_std,
    color= "tab:blue",
    alpha = 0.15,
    label = "GP 95% CI"
)

plt.grid(True, linestyle= ":", alpha = 0.6)
plt.tight_layout()

plt.show()
```
