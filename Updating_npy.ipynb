```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
cwd=Path.cwd()
folder= "results_v14/results/raw_data"
raw_paths= list((cwd/folder).glob("*.npy"))

file_paths= sorted(
    raw_paths,
    key= lambda p: int(p.stem.split("_")[2])
)

iter_list=[np.load(file,allow_pickle=True).item() for file in file_paths]
```

```python
saving_name= "raw_run_"
for i,value in enumerate(iter_list):
    value['metadata']["unit"] = "Pa"
    np.save(cwd/folder/f"{saving_name}{i}",value)
```
